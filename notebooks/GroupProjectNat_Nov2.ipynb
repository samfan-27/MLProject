```python
import sys, os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath(os.path.join('..', 'scripts')))
from data_loader import create_panel_data

panel_df = create_panel_data(frequency='5_min')

print('Data Head: ')
print(panel_df.head())
print('\nData Tail: ')
print(panel_df.tail())

START_DATE = '2003-01-02'
END_DATE = '2024-03-28'

vix_path = os.path.join('..', 'data', 'VIX_History.csv')
vix_data = pd.read_csv(vix_path)

vix_data['Date'] = pd.to_datetime(vix_data['DATE'], format='%m/%d/%Y')
vix_data = vix_data.set_index('Date')
vix = vix_data[['CLOSE']].rename(columns={'CLOSE': 'vix'})
vix = vix.loc[START_DATE:END_DATE]

panel_vix = panel_df.join(vix, on='Date')
print('Panel with VIX: ')
print(panel_vix.head())

df = panel_vix.sort_index()

#df['log_rv'] = np.log(df['rv'])
#df['Y_reg'] = df.groupby('Stock')['log_rv'].shift(-1) # The new target name (Y_reg)


df['Y_reg'] = df.groupby('Stock')['rv'].shift(-1)

print('Example for AAPL: ')
display(df.loc[pd.IndexSlice[:, 'AAPL'], :].tail()[['rv', 'Y_reg']])

# Engineering HAR Features
#changed from rv to Log_rv
df['rv_lag_1'] = df.groupby('Stock')['rv'].shift(1)
df['rv_rolling_5'] = df.groupby('Stock')['rv_lag_1'].rolling(window=5, min_periods=1).mean().reset_index(level=0, drop=True)
df['rv_rolling_22'] = df.groupby('Stock')['rv_lag_1'].rolling(window=22, min_periods=1).mean().reset_index(level=0, drop=True)

df['bpv_lag_1'] = df.groupby('Stock')['bpv'].shift(1)
df['good_lag_1'] = df.groupby('Stock')['good'].shift(1)
df['bad_lag_1'] = df.groupby('Stock')['bad'].shift(1)
df['rq_lag_1'] = df.groupby('Stock')['rq'].shift(1)
df['vix_lag_1'] = df.groupby('Stock')['vix'].shift(1)

epsilon = 1e-10
df['bad_good_ratio_lag_1'] = df['bad_lag_1'] / (df['good_lag_1'] + epsilon)
df['jump_ratio_lag_1'] = (df['rv_lag_1'] - df['bpv_lag_1']) / (df['rv_lag_1'] + epsilon)
display(df.head())
display(df.tail())


# Additional Feature Engineering
print("=== Additional Feature Engineering ===\n")

# Volatility ratios and differences
#changed to log_rv
df['rv_bpv_ratio'] = df['rv_lag_1'] / (df['bpv_lag_1'] + epsilon)
df['good_bad_diff'] = df['good_lag_1'] - df['bad_lag_1']
df['rv_change'] = df['rv_lag_1'] - df.groupby('Stock')['rv'].shift(2)

# VIX interaction terms
df['rv_vix_interaction'] = df['rv_lag_1'] * df['vix_lag_1']

# Volatility of volatility (rolling std)
df['rv_rolling_std_5'] = df.groupby('Stock')['rv_lag_1'].rolling(window=5, min_periods=1).std().reset_index(level=0, drop=True)
df['rv_rolling_std_22'] = df.groupby('Stock')['rv_lag_1'].rolling(window=22, min_periods=1).std().reset_index(level=0, drop=True)

print("Feature engineering complete!")
print(f"Total features: {df.shape[1]}")
print(f"\nNew features added: rv_bpv_ratio, good_bad_diff, rv_change, rv_vix_interaction, rv_rolling_std_5, rv_rolling_std_22")

# ============================================================================
# 1. HAR BASELINE MODEL
# ============================================================================
print("\n" + "="*80)
print("=== Building HAR Baseline Model ===")
print("="*80 + "\n")

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Prepare data for HAR model
har_features = ['rv_lag_1', 'rv_rolling_5', 'rv_rolling_22']
df_har = df[har_features + ['Y_reg']].dropna().copy()

# Add Date and Stock from the index
df_har['Date'] = df_har.index.get_level_values('Date')
df_har['Stock'] = df_har.index.get_level_values('Stock')

# Split data: use 2003-2022 for training, 2023-2024 for testing
train_har = df_har[df_har['Date'] < '2023-01-01']
test_har = df_har[df_har['Date'] >= '2023-01-01']

X_train_har = train_har[har_features]
y_train_har = train_har['Y_reg']
X_test_har = test_har[har_features]
y_test_har = test_har['Y_reg']

# Train HAR model
har_model = LinearRegression()
har_model.fit(X_train_har, y_train_har)

# Make predictions
y_pred_har_train = har_model.predict(X_train_har)
y_pred_har_test = har_model.predict(X_test_har)

# Calculate metrics
train_mse_har = mean_squared_error(y_train_har, y_pred_har_train)
train_mae_har = mean_absolute_error(y_train_har, y_pred_har_train)
train_r2_har = r2_score(y_train_har, y_pred_har_train)

test_mse_har = mean_squared_error(y_test_har, y_pred_har_test)
test_mae_har = mean_absolute_error(y_test_har, y_pred_har_test)
test_r2_har = r2_score(y_test_har, y_pred_har_test)

print("HAR Model Coefficients:")
for feature, coef in zip(har_features, har_model.coef_):
    print(f"  {feature}: {coef:.4f}")
print(f"  Intercept: {har_model.intercept_:.4f}")

print(f"\nTrain Metrics:")
print(f"  MSE: {train_mse_har:.4f}")
print(f"  MAE: {train_mae_har:.4f}")
print(f"  R²: {train_r2_har:.4f}")

print(f"\nTest Metrics:")
print(f"  MSE: {test_mse_har:.4f}")
print(f"  MAE: {test_mae_har:.4f}")
print(f"  R²: {test_r2_har:.4f}")

# ============================================================================
# 2. ISOLATION FOREST FOR ANOMALY DETECTION
# ============================================================================
print("\n" + "="*80)
print("=== Detecting Anomalies with Isolation Forest ===")
print("="*80 + "\n")

from sklearn.ensemble import IsolationForest

# Calculate prediction errors on training set
train_har['residual'] = y_train_har - y_pred_har_train
train_har['abs_residual'] = np.abs(train_har['residual'])

# Use Isolation Forest on residuals
iso_forest = IsolationForest(contamination=0.05, random_state=42)
train_har['is_abnormal'] = iso_forest.fit_predict(train_har[['abs_residual']].values)
train_har['is_abnormal'] = (train_har['is_abnormal'] == -1).astype(int)

# Apply to test set
test_har['residual'] = y_test_har - y_pred_har_test
test_har['abs_residual'] = np.abs(test_har['residual'])
test_har['is_abnormal'] = iso_forest.predict(test_har[['abs_residual']].values)
test_har['is_abnormal'] = (test_har['is_abnormal'] == -1).astype(int)

anomaly_pct_train = train_har['is_abnormal'].mean() * 100
anomaly_pct_test = test_har['is_abnormal'].mean() * 100

print(f"Anomalies detected in training set: {anomaly_pct_train:.2f}%")
print(f"Anomalies detected in test set: {anomaly_pct_test:.2f}%")

# Merge anomaly labels back to original dataframe
df_har_full = pd.concat([train_har, test_har])
df = df.join(df_har_full[['is_abnormal', 'residual']], how='left')

# ============================================================================
# 3. XGBOOST REGRESSION MODEL
# ============================================================================
print("\n" + "="*80)
print("=== Building XGBoost Regression Model ===")
print("="*80 + "\n")

import xgboost as xgb

# Select features for XGBoost
xgb_features = ['rv_lag_1', 'rv_rolling_5', 'rv_rolling_22', 
                'bpv_lag_1', 'good_lag_1', 'bad_lag_1', 'rq_lag_1',
                'vix_lag_1', 'bad_good_ratio_lag_1', 'jump_ratio_lag_1',
                'rv_bpv_ratio', 'good_bad_diff', 'rv_change', 
                'rv_vix_interaction', 'rv_rolling_std_5', 'rv_rolling_std_22']

df_xgb = df[xgb_features + ['Y_reg']].dropna().copy()
df_xgb['Date'] = df_xgb.index.get_level_values('Date')
df_xgb['Stock'] = df_xgb.index.get_level_values('Stock')

train_xgb = df_xgb[df_xgb['Date'] < '2023-01-01']
test_xgb = df_xgb[df_xgb['Date'] >= '2023-01-01']

X_train_xgb = train_xgb[xgb_features]
y_train_xgb = train_xgb['Y_reg']
X_test_xgb = test_xgb[xgb_features]
y_test_xgb = test_xgb['Y_reg']

# Train XGBoost regression model
xgb_reg = xgb.XGBRegressor(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.1,
    random_state=42
)
xgb_reg.fit(X_train_xgb, y_train_xgb)

# Make predictions
y_pred_xgb_train = xgb_reg.predict(X_train_xgb)
y_pred_xgb_test = xgb_reg.predict(X_test_xgb)

# Calculate metrics
train_mse_xgb = mean_squared_error(y_train_xgb, y_pred_xgb_train)
train_mae_xgb = mean_absolute_error(y_train_xgb, y_pred_xgb_train)
train_r2_xgb = r2_score(y_train_xgb, y_pred_xgb_train)

test_mse_xgb = mean_squared_error(y_test_xgb, y_pred_xgb_test)
test_mae_xgb = mean_absolute_error(y_test_xgb, y_pred_xgb_test)
test_r2_xgb = r2_score(y_test_xgb, y_pred_xgb_test)

print(f"Train Metrics:")
print(f"  MSE: {train_mse_xgb:.4f}")
print(f"  MAE: {train_mae_xgb:.4f}")
print(f"  R²: {train_r2_xgb:.4f}")

print(f"\nTest Metrics:")
print(f"  MSE: {test_mse_xgb:.4f}")
print(f"  MAE: {test_mae_xgb:.4f}")
print(f"  R²: {test_r2_xgb:.4f}")

# ============================================================================
# 4. XGBOOST CLASSIFICATION MODEL
# ============================================================================
print("\n" + "="*80)
print("=== Building XGBoost Classification Model ===")
print("="*80 + "\n")

from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Prepare data with anomaly labels
df_clf = df[xgb_features + ['is_abnormal']].dropna().copy()
df_clf['Date'] = df_clf.index.get_level_values('Date')
df_clf['Stock'] = df_clf.index.get_level_values('Stock')

train_clf = df_clf[df_clf['Date'] < '2023-01-01']
test_clf = df_clf[df_clf['Date'] >= '2023-01-01']

X_train_clf = train_clf[xgb_features]
y_train_clf = train_clf['is_abnormal']
X_test_clf = test_clf[xgb_features]
y_test_clf = test_clf['is_abnormal']

# Train XGBoost classifier
xgb_clf = xgb.XGBClassifier(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.1,
    random_state=42,
    scale_pos_weight=len(y_train_clf[y_train_clf==0])/len(y_train_clf[y_train_clf==1])
)
xgb_clf.fit(X_train_clf, y_train_clf)

# Make predictions
y_pred_clf_train = xgb_clf.predict(X_train_clf)
y_pred_clf_test = xgb_clf.predict(X_test_clf)
y_pred_proba_test = xgb_clf.predict_proba(X_test_clf)[:, 1]

# Calculate metrics
train_auc = roc_auc_score(y_train_clf, xgb_clf.predict_proba(X_train_clf)[:, 1])
test_auc = roc_auc_score(y_test_clf, y_pred_proba_test)

print(f"Train AUC: {train_auc:.4f}")
print(f"Test AUC: {test_auc:.4f}")

print("\nTest Set Classification Report:")
print(classification_report(y_test_clf, y_pred_clf_test, target_names=['Normal', 'Abnormal']))

# ============================================================================
# 5. VISUALIZATIONS
# ============================================================================
print("\n" + "="*80)
print("=== Creating Visualizations ===")
print("="*80 + "\n")

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Model Comparison
ax1 = axes[0, 0]
models = ['HAR\nBaseline', 'XGBoost\nRegression']
test_mse_values = [test_mse_har, test_mse_xgb]
test_mae_values = [test_mae_har, test_mae_xgb]

x = np.arange(len(models))
width = 0.35

bars1 = ax1.bar(x - width/2, test_mse_values, width, label='MSE', alpha=0.8)
bars2 = ax1.bar(x + width/2, test_mae_values, width, label='MAE', alpha=0.8)

ax1.set_ylabel('Error', fontsize=12)
ax1.set_title('Model Comparison: Test Set Performance', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(models)
ax1.legend()
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.3f}', ha='center', va='bottom', fontsize=10)

# 2. Feature Importance (Top 10)
ax2 = axes[0, 1]
importance = xgb_reg.feature_importances_
feature_importance = pd.DataFrame({
    'feature': xgb_features,
    'importance': importance
}).sort_values('importance', ascending=False).head(10)

ax2.barh(range(len(feature_importance)), feature_importance['importance'], alpha=0.8)
ax2.set_yticks(range(len(feature_importance)))
ax2.set_yticklabels(feature_importance['feature'])
ax2.set_xlabel('Importance', fontsize=12)
ax2.set_title('Top 10 Feature Importance (XGBoost)', fontsize=14, fontweight='bold')
ax2.invert_yaxis()
ax2.grid(axis='x', alpha=0.3)

# 3. Actual vs Predicted (sample stock)
ax3 = axes[1, 0]
sample_stock = 'AAPL'
sample_data = test_xgb[test_xgb['Stock'] == sample_stock].copy()
sample_data['predicted'] = xgb_reg.predict(sample_data[xgb_features])

dates = sample_data['Date'].values[-100:]  # Last 100 days
actual = sample_data['Y_reg'].values[-100:]
predicted = sample_data['predicted'].values[-100:]

ax3.plot(dates, actual, label='Actual', alpha=0.7, linewidth=2)
ax3.plot(dates, predicted, label='Predicted', alpha=0.7, linewidth=2)
ax3.set_xlabel('Date', fontsize=12)
ax3.set_ylabel('Volatility', fontsize=12)
ax3.set_title(f'Actual vs Predicted Volatility: {sample_stock}', fontsize=14, fontweight='bold')
ax3.legend()
ax3.grid(alpha=0.3)
ax3.tick_params(axis='x', rotation=45)

# 4. Anomaly Detection Distribution
ax4 = axes[1, 1]
normal_residuals = train_har[train_har['is_abnormal'] == 0]['abs_residual']
abnormal_residuals = train_har[train_har['is_abnormal'] == 1]['abs_residual']

ax4.hist(normal_residuals, bins=50, alpha=0.6, label='Normal', density=True)
ax4.hist(abnormal_residuals, bins=50, alpha=0.6, label='Abnormal', density=True)
ax4.set_xlabel('Absolute Residual', fontsize=12)
ax4.set_ylabel('Density', fontsize=12)
ax4.set_title('Distribution of Prediction Errors', fontsize=14, fontweight='bold')
ax4.legend()
ax4.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Additional: Confusion Matrix for Classification
fig, ax = plt.subplots(1, 1, figsize=(8, 6))
cm = confusion_matrix(y_test_clf, y_pred_clf_test)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
            xticklabels=['Normal', 'Abnormal'],
            yticklabels=['Normal', 'Abnormal'])
ax.set_ylabel('Actual', fontsize=12)
ax.set_xlabel('Predicted', fontsize=12)
ax.set_title('Confusion Matrix: Anomaly Classification', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ============================================================================
# 6. MODEL COMPARISON SUMMARY
# ============================================================================
print("\n" + "="*80)
print("=== MODEL COMPARISON SUMMARY ===")
print("="*80 + "\n")

comparison_df = pd.DataFrame({
    'Model': ['HAR Baseline', 'XGBoost Regression'],
    'Test MSE': [test_mse_har, test_mse_xgb],
    'Test MAE': [test_mae_har, test_mae_xgb],
    'Test R²': [test_r2_har, test_r2_xgb]
})

print(comparison_df.to_string(index=False))

improvement_mse = ((test_mse_har - test_mse_xgb) / test_mse_har) * 100
improvement_mae = ((test_mae_har - test_mae_xgb) / test_mae_har) * 100

print(f"\nXGBoost Improvement over HAR:")
print(f"  MSE: {improvement_mse:.2f}% reduction")
print(f"  MAE: {improvement_mae:.2f}% reduction")

print("\n" + "="*80)
print("Analysis Complete!")
print("="*80)


# ============================================================================
# 7. HIDDEN MARKOV MODEL - REGIME DETECTION
# ============================================================================
print("\n" + "="*80)
print("=== Hidden Markov Model: Volatility Regime Detection ===")
print("="*80 + "\n")

from hmmlearn import hmm

# Select a stock for regime analysis
hmm_stock = 'AAPL'
print(f"Analyzing volatility regimes for: {hmm_stock}\n")

# Prepare data for HMM
hmm_data = df.loc[pd.IndexSlice[:, hmm_stock], ['rv', 'bpv', 'good', 'bad']].dropna()
X_hmm = hmm_data.values

# Fit Gaussian HMM with 3 regimes (Low, Medium, High volatility)
n_regimes = 3
model_hmm = hmm.GaussianHMM(n_components=n_regimes, covariance_type="full", 
                             n_iter=1000, random_state=42)
model_hmm.fit(X_hmm)

# Predict regimes
regimes = model_hmm.predict(X_hmm)
hmm_data['regime'] = regimes

# Label regimes based on average RV in each regime
regime_stats = hmm_data.groupby('regime')['rv'].mean().sort_values()
regime_mapping = {regime_stats.index[0]: 'Low', 
                  regime_stats.index[1]: 'Medium', 
                  regime_stats.index[2]: 'High'}
hmm_data['regime_label'] = hmm_data['regime'].map(regime_mapping)

# Get anomaly labels for comparison
hmm_data = hmm_data.join(df.loc[pd.IndexSlice[:, hmm_stock], ['is_abnormal']], how='left')

print("Regime Statistics:")
print("-" * 60)
regime_summary = hmm_data.groupby('regime_label')['rv'].agg(['count', 'mean', 'std', 'min', 'max'])
print(regime_summary)

# Compare regimes with anomalies
print("\n\nRegime vs Anomaly Cross-tabulation:")
print("-" * 60)
crosstab = pd.crosstab(hmm_data['regime_label'], 
                       hmm_data['is_abnormal'].fillna(0), 
                       normalize='index') * 100
crosstab.columns = ['Normal (%)', 'Abnormal (%)']
print(crosstab.round(2))

# ============================================================================
# 8. HMM VISUALIZATIONS
# ============================================================================
print("\n" + "="*80)
print("=== Creating HMM Visualizations ===")
print("="*80 + "\n")

fig, axes = plt.subplots(3, 1, figsize=(15, 12))

# 1. Time Series with Regimes
ax1 = axes[0]
dates_hmm = hmm_data.index.get_level_values('Date')

# Color map for regimes
colors = {'Low': 'green', 'Medium': 'orange', 'High': 'red'}
for regime_name in ['Low', 'Medium', 'High']:
    mask = hmm_data['regime_label'] == regime_name
    ax1.scatter(dates_hmm[mask], hmm_data['rv'][mask], 
               c=colors[regime_name], label=regime_name, alpha=0.6, s=10)

ax1.set_xlabel('Date', fontsize=12)
ax1.set_ylabel('Realized Volatility', fontsize=12)
ax1.set_title(f'Volatility Regimes Over Time: {hmm_stock}', fontsize=14, fontweight='bold')
ax1.legend(title='Regime', loc='upper right')
ax1.grid(alpha=0.3)

# 2. Regime Distribution
ax2 = axes[1]
regime_counts = hmm_data['regime_label'].value_counts().reindex(['Low', 'Medium', 'High'])
bars = ax2.bar(regime_counts.index, regime_counts.values, 
               color=[colors[r] for r in regime_counts.index], alpha=0.7)
ax2.set_ylabel('Number of Days', fontsize=12)
ax2.set_title('Distribution of Volatility Regimes', fontsize=14, fontweight='bold')
ax2.grid(axis='y', alpha=0.3)

# Add percentage labels
total = regime_counts.sum()
for bar, count in zip(bars, regime_counts.values):
    height = bar.get_height()
    pct = (count / total) * 100
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(count)}\n({pct:.1f}%)',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# 3. Regime vs Anomaly Comparison
ax3 = axes[2]
# Prepare data for last 2 years
recent_data = hmm_data[hmm_data.index.get_level_values('Date') >= '2022-01-01'].copy()
dates_recent = recent_data.index.get_level_values('Date')

# Plot RV
ax3_twin = ax3.twinx()
ax3.plot(dates_recent, recent_data['rv'], color='blue', alpha=0.5, linewidth=1.5, label='RV')

# Mark anomalies
anomaly_mask = recent_data['is_abnormal'] == 1
ax3.scatter(dates_recent[anomaly_mask], recent_data['rv'][anomaly_mask], 
           color='purple', s=100, marker='x', linewidths=3, label='Anomaly', zorder=5)

# Color background by regime
for regime_name, color in colors.items():
    regime_mask = recent_data['regime_label'] == regime_name
    if regime_mask.any():
        # Create filled regions for regimes
        regime_dates = dates_recent[regime_mask]
        regime_rv = recent_data['rv'][regime_mask]
        ax3.scatter(regime_dates, [-0.5] * len(regime_dates), 
                   c=color, alpha=0.3, s=100, marker='s', label=f'{regime_name} Regime')

ax3.set_xlabel('Date', fontsize=12)
ax3.set_ylabel('Realized Volatility', fontsize=12)
ax3.set_title(f'Regimes vs Anomalies (Recent Period): {hmm_stock}', fontsize=14, fontweight='bold')
ax3.legend(loc='upper left', fontsize=10)
ax3.grid(alpha=0.3)
ax3.set_ylim(bottom=-1)

plt.tight_layout()
plt.show()

# ============================================================================
# 9. REGIME TRANSITION MATRIX
# ============================================================================
print("\n" + "="*80)
print("=== Regime Transition Analysis ===")
print("="*80 + "\n")

# Calculate transition probabilities
transition_matrix = model_hmm.transmat_
regime_names = ['Low', 'Medium', 'High']

# Map to sorted regime names
sorted_indices = regime_stats.index
sorted_names = [regime_mapping[i] for i in sorted_indices]

print("Transition Probability Matrix:")
print("-" * 60)
transition_df = pd.DataFrame(transition_matrix, 
                             index=sorted_names,
                             columns=sorted_names)
print(transition_df.round(3))

# Visualize transition matrix
fig, ax = plt.subplots(1, 1, figsize=(8, 6))
sns.heatmap(transition_df, annot=True, fmt='.3f', cmap='YlOrRd', 
            ax=ax, cbar_kws={'label': 'Probability'}, vmin=0, vmax=1)
ax.set_title('Regime Transition Probability Matrix', fontsize=14, fontweight='bold')
ax.set_xlabel('To Regime', fontsize=12)
ax.set_ylabel('From Regime', fontsize=12)
plt.tight_layout()
plt.show()

# ============================================================================
# 10. KEY INSIGHTS
# ============================================================================
print("\n" + "="*80)
print("=== KEY INSIGHTS ===")
print("="*80 + "\n")

# 1. Model Performance
print("1. MODEL PERFORMANCE:")
print(f"   - HAR Baseline Test MSE: {test_mse_har:.4f}")
print(f"   - XGBoost Regression Test MSE: {test_mse_xgb:.4f}")
print(f"   - Improvement: {improvement_mse:.2f}%")

# 2. Anomaly Detection
print("\n2. ANOMALY DETECTION:")
print(f"   - Anomalies identified: {anomaly_pct_test:.2f}% of test days")
print(f"   - Classification AUC: {test_auc:.4f}")

# 3. Regime Analysis
high_regime_pct = (regime_counts['High'] / total) * 100
print(f"\n3. REGIME ANALYSIS ({hmm_stock}):")
print(f"   - High volatility regime: {high_regime_pct:.1f}% of days")
print(f"   - Most stable regime: {sorted_names[0]} (highest self-transition: {transition_matrix[0,0]:.3f})")

# 4. Regime-Anomaly Connection
high_regime_data = hmm_data[hmm_data['regime_label'] == 'High']
high_regime_anomaly_pct = (high_regime_data['is_abnormal'].sum() / len(high_regime_data)) * 100
print(f"   - Anomalies in high volatility regime: {high_regime_anomaly_pct:.1f}%")

print("\n" + "="*80)
print("Complete Analysis Finished!")
print("="*80)
```
